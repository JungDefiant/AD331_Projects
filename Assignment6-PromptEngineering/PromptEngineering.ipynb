```python
### Prompt Engineering for Performance Improvement
from transformers import AutoTokenizer, AutoModelForCausalLM
model_name = 'distilgpt2'
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name, output_hidden_states=True)
```

```python
text = "Finish the following. DM: Galstaff, you enter the door to the north."
input = tokenizer.encode(text, return_tensors="pt")
output = model.generate(
    input, 
    max_length=400,
    do_sample=True, pad_token_id=tokenizer.eos_token_id
)
output_text1 = tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
## Iterative Prompt Engineering
# Role Prompting
text = "Pretend to be a Dungeon Master.\nFinish the following. DM: Galstaff, you enter the door to the north."
input = tokenizer.encode(text, return_tensors="pt")
output = model.generate(
    input, 
    max_length=400, 
    do_sample=True, pad_token_id=tokenizer.eos_token_id
)
output_text2 = tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
# Output Formatting
text = "Use the following format. \nDM: <Describes what happens>\nPlayer: <Reacts to what happens>\nFinish the following. DM: Galstaff, you enter the door to the north."
input = tokenizer.encode(text, return_tensors="pt")
output = model.generate(
    input, 
    max_length=400, 
    do_sample=True, pad_token_id=tokenizer.eos_token_id
)
output_text3 = tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
# One-Shot Prompting
text = "DM: You've been following this tunnel for about 120 yards. The water on the floor is ankle deep and very cold. Now and then you feel something brush against your foot. The smell of decay is getting stronger. The tunnel is gradually filling with a cold mist.\nFighter: I don't like this at all. Can we see anything up ahead that looks like a doorway, or a branch in the tunnel?\nDM: Within the range of your torchlight, the tunnel is more or less straight. You don't see any branches or doorways.\nCleric: The wererat we hit had to come this way. There's nowhere else to go.\nFighter: Unless we missed a hidden door along the way. I hate this place; it gives me the creeps.\nFinish the following using the prior example. DM: Galstaff, you enter the door to the north."
input = tokenizer.encode(text, return_tensors="pt")
output = model.generate(
    input, 
    max_length=400,
    do_sample=True, pad_token_id=tokenizer.eos_token_id
)
output_text4 = tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
## Qualitative Evaluation
import textwrap
print(textwrap.fill(output_text1.replace("\n", " "), width=100))
print(f"\n{textwrap.fill("EVAL: This output is largely gibberish that is somewhat related to the original prompt. The model " \
"seems to somewhat understand the context of being realted to an RPG, but it doesn't understand what I'm " \
"trying to do, which is emulate an interaction in a tabletop roleplaying game.", width=100)}")
```

```python
print(textwrap.fill(output_text2.replace("\n", " "), width=100))
print(f"\n{textwrap.fill("EVAL: This output seems to be very confused, even if it understands the context " \
"of emulating an RPG. It seems to think that it's laying out a design for a dungeon, not pretending to " \
"interact with players in a tabletop game.", width=100)}")
```

```python
print(textwrap.fill(output_text3.replace("\n", " "), width=100))
print(f"\n{textwrap.fill("EVAL: This output seems to think that it's describing general concepts of " \
"an MMO or an RPG, but it's mostly nonsense and makes me think of a conversation in a dream. And then " \
"it tries selling a fake service.", width=100)}")
```

```python
print(textwrap.fill(output_text4.replace("\n", " "), width=100))
print(f"\n{textwrap.fill("This output is the most nonsensical and confused. It doesn't seem to understand " \
"how to adapt the example given.", width=100)}")
```

```python
## Final Prompt
text = "Pretend to be a Dungeon Master. Use the following format. \nDM: <Describes what happens>\nPlayer: <Reacts to what happens>\nFinish the following. DM: Galstaff, you enter the door to the north."
input = tokenizer.encode(text, return_tensors="pt")
output = model.generate(
    input, 
    max_length=400,
    do_sample=True, pad_token_id=tokenizer.eos_token_id
)
final_output_text = tokenizer.decode(output[0], skip_special_tokens=True)

```
